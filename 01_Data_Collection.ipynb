```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("literacy_analysis.db")
```

```python
query = """
SELECT Country, adult_literacy_rate
FROM literacy_rates
WHERE Year = 2020
ORDER BY adult_literacy_rate DESC
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Country, youth_literacy_female
FROM literacy_rates
WHERE youth_literacy_female < 80
ORDER BY youth_literacy_female;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT i.world_region,
       AVG(l.adult_literacy_rate) AS avg_adult_literacy
FROM literacy_rates l
JOIN illiteracy_population i
ON l.Country = i.Country
AND l.Year = i.Year
GROUP BY i.world_region
ORDER BY avg_adult_literacy DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Country, illiteracy_rate
FROM illiteracy_population
WHERE Year = 2000
AND illiteracy_rate > 20
ORDER BY illiteracy_rate DESC;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Year, illiteracy_rate
FROM illiteracy_population
WHERE Country = 'India'
AND Year BETWEEN 2000 AND 2020
ORDER BY Year;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Country, Year, illiteracy_rate
FROM illiteracy_population
WHERE Year = (SELECT MAX(Year) FROM illiteracy_population)
ORDER BY illiteracy_rate DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Country, avg_years_schooling, gdp_per_capita
FROM gdp_schooling
WHERE avg_years_schooling > 7
AND gdp_per_capita < 5000;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Country,
       gdp_per_capita,
       RANK() OVER (ORDER BY gdp_per_capita DESC) AS GDP_Rank
FROM gdp_schooling
WHERE Year = 2020;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Year,
       AVG(avg_years_schooling) AS Global_Avg_Schooling
FROM gdp_schooling
GROUP BY Year
ORDER BY Year;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Country, gdp_per_capita, avg_years_schooling
FROM gdp_schooling
WHERE Year = 2020
AND avg_years_schooling < 6
ORDER BY gdp_per_capita DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT g.Country,
       g.avg_years_schooling,
       i.illiteracy_rate
FROM gdp_schooling g
JOIN illiteracy_population i
ON g.Country = i.Country
AND g.Year = i.Year
WHERE g.avg_years_schooling > 10
AND i.illiteracy_rate > 20;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT l.Year,
       l.literacy_rate,
       g.gdp_per_capita
FROM literacy_rates l
JOIN gdp_schooling g
ON l.Country = g.Country
AND l.Year = g.Year
WHERE l.Country = 'India'
AND l.Year BETWEEN 2000 AND 2020
ORDER BY l.Year;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT l.Country,
       l.youth_literacy_male,
       l.youth_literacy_female,
       (l.youth_literacy_male - l.youth_literacy_female) AS literacy_gap,
       g.gdp_per_capita
FROM literacy_rates l
JOIN gdp_schooling g
ON l.Country = g.Country
AND l.Year = g.Year
WHERE g.Year = 2020
AND g.gdp_per_capita > 30000
ORDER BY literacy_gap DESC;
"""

pd.read_sql_query(query, conn)
```
